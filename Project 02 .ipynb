```python
# Import lớp Image
from PIL import Image
from IPython.display import display
from imgtools import *
```

```python
# Thư mục chứa hình ảnh
my_directory = "./user/folder_image" #Đường dẫn này là đường dẫn thư mục(folder) => nó sẽ in ra tất cả hình ảnh có trong thư mục.
```

```python
# Đọc nhiều ảnh
images = get_image_list(my_directory) #điền vào my_directory là thư mục(folder) mới chạy được code.
```

```python
# Hiển thị toàn bộ ảnh bằng hàm display
for img in images:
    print(img.size)
    display(img)
```
