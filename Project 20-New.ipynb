```python
import cv2
import numpy as np 
```

```python
def display(title, img):
    cv2.imshow(title, img)
    # Chờ một khoảng thời gian
    cv2.waitKey(0) # có thể cho thời gian mili giây khác(đặt thời gian cho tab => tự động tắt tab!)
## Đóng tất cả cửa sổ tab
    cv2.destroyWindow(title)
```

```python
# Tạo ra một hình ảnh
img = np.zeros((400,400,3), dtype=np.uint8)
display('Image',img)
```

```python
# Vẽ đường thẳng
 #Thông số bên trong line: (image, start_point, end_point, color, thickness)
cv2.line(img, (50,50), (350,50), (0,0,255),2)
display('Image', img)
```

```python
# Vẽ hình tròn
 #Thông số bên trong circle: (image, center, radius, color, thickness)
cv2.circle(img, (200,200), 50, (0,255,0), -1)
display('Image', img)
```

```python
# Vẽ hình chữ nhật 
 #Thông số bên trong rectangle: (image, top_left_point, bottom_right_point, color, thickness)
cv2.rectangle(img, (100,100),(300,300), (255,0,0),-1)
display('Image', img)
```

```python
# Vẽ hình đa giác
 #Thông số bên trong polylines: (image, [array_of_points], is_closed, color, thickness)
pts = np.array([[200,50], [300,150], [250,300], [150,300], [100,150]], np.int32)
pts = pts.reshape((-1,1,2))
cv2.polylines(img, [pts], True, (255,0,0), 2)
display('Image', img)
```

```python
# Vẽ hình tam giác
pts = np.array([[100,100], [200,50], [300,100]], np.int32)
pts = pts.reshape((-1,1,2))
cv2.polylines(img, [pts], True, (255,0,0), 2)
display('Image', img)
```

```python
# Vẽ hình tứ giác (chữ nhật)
pts = np.array([[100,100], [300,100], [300,300], [100,300]], np.int32)
pts = pts.reshape((-1,1,2))
cv2.polylines(img, [pts], True, (0,255,0), 2)
display('Image', img)
```

```python
# Vẽ hình ngũ giác
pts = np.array([[200,50], [300,150], [250,300], [150,300], [100,150]], np.int32)
pts = pts.reshape((-1,1,2))
cv2.polylines(img, [pts], True, (0,0,255), 2)
display('Image', img)
```

```python
# Vẽ hình ellipse
 #Thông số bên trong ellipse: (image, center, axes, angle, start_angle, end_angle, color, thickness)
cv2.ellipse(img, (50,50),(20,20),0,0,360, (255,0,0),-1)
display('Image', img)
```

```python
# Ghi text trên hình ảnh
my_content = 'Do Nguyen Tin Tin'
font = cv2.FONT_HERSHEY_SIMPLEX
# Thông số bên trong text: (image, text, org(tọa độ điểm góc dưới bên trái của chữ), font, fontscale(hệ số phóng to chữ), color, thickness) 
cv2.putText(img, my_content, (10,380), font, 0.6, (0,0,255),2)
display('Image', img)
```
