```python
# Giải hệ phương trình tuyến tính
from scipy import linalg # linalg viết tắt của Linear Algebra
import numpy as np

# Định nghĩa hệ phương trình tuyến tính
A = np.array([[2, 1],[3, 2]])
B = np.array([5,7])
#Giải thích: 2x + y = 5 và 3x + 2y = 7

# Giải phương trình tuyến tính 
x = linalg.solve(A, B)
print("Kết quả: ", x)
```

```python
# Tính tích phân của một hàm số
from scipy import integrate 

# Định nghĩa hàm f(x) = x^2
def my_function(x):
    return x**2

# Tính tích phân của f(x) từ 0 đến 1
integral = integrate.quad(my_function, 0, 1); # trong toán học, “quadrature” có nghĩa là tính tích phân (numerical integration).
print(integral)
print("Vậy tích phân f(x) từ 0 đến 1: ", integral)
```

```python
from scipy import linalg
import numpy as np 
# Tính giá trị riêng và vector riêng của một ma trận
A = np.array([[2,1], [3,2]])

# Tính giá trị riêng và vector riêng của ma trận
evals, evecs = linalg.eig(A)

print("Giá trị riêng: ", evals) # evals là giá trị riêng (eigenvalues)
print("Vector riêng: ", evecs) # evecs là vector riêng (eigenvectors)
```
