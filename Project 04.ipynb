```python
# Import lớp Image
from PIL import Image
from imgtools import *
from IPython.display import display #để hiển thị ra hình ảnh
```

```python
# Đường dẫn hình ảnh:
my_path = 'image.img'
```

```python
# Đọc hình ảnh
img = load_image(my_path)
```

```python
# Hiển thị hình ảnh ra màn hình
display(img)
```

```python
# Định vị khu vực cắt
region = (50,50,700,1200) # Thông số này là (left, upper, right, lower)
# Cắt vùng hình ảnh
cropped_image = img.crop(region)
display(cropped_image)
```

```python
# Dán hình ảnh
# Xác định vị trí cần dán
paste_position = (150,150) 
img.paste(cropped_image, paste_position)
display(img)
```
