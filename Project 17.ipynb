```python
# Bộ Lọc Sobel (Sobel Filter)
from PIL import Image
import numpy as np 
import matplotlib.pyplot as plt 
from scipy import ndimage

# Đọc hình ảnh
image_path = 'BCO.dea463a3-a6c8-48ee-b62a-054707359971.png'
img = Image.open(image_path)

# Chuyển sang ảnh xám
gray_image = img.convert('L')

#Chuyển ảnh xám về mảng
image_array = np.array(gray_image)

# Tính gradient(độ dốc) theo hướng x 
gradient_x = ndimage.sobel(image_array, axis=0, mode='constant') # Giải thích: axis=0 => theo hướng x, axis=1 => theo hướng y.
# Tính gradient(độ dốc) theo hướng y
gradient_y = ndimage.sobel(image_array, axis=1, mode='constant') # Giải thích: axis=0 => theo hướng x, axis=1 => theo hướng y.

# Hiển thị ảnh
fig, axs = plt.subplots(3,1, figsize = (13,18)) # Giải thích: 2 là số hàng, 1 là số cột

plt.subplot(3,1,1)
plt.imshow(gray_image, cmap='gray')
plt.title('Ảnh xám gốc')

plt.subplot(3,1,2)
plt.imshow(gradient_x, cmap='gray')
plt.title('Ảnh xám theo gradient x')

plt.subplot(3,1,3)
plt.imshow(gradient_y, cmap='gray')
plt.title('Ảnh xám theo gradient y')

plt.show()
```

```python
# Bộ lọc Prewitt (Prewitt Filter)
from PIL import Image
import numpy as np 
import matplotlib.pyplot as plt 
from scipy import ndimage

# Đọc hình ảnh
image_path = 'BCO.dea463a3-a6c8-48ee-b62a-054707359971.png'
img = Image.open(image_path)

# Chuyển sang ảnh xám
gray_image = img.convert('L')

#Chuyển ảnh xám về mảng
image_array = np.array(gray_image)

# Tính gradient theo hướng x (đạo hàm riêng theo x) bằng mặt nạ Prewitt x
gradient_x = ndimage.convolve(image_array, np.array([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]])) # Ý nghĩa: -1 → giảm cường độ của pixel bên trái, 0 → pixel trung tâm không tác động, 1 → tăng cường độ pixel bên phải. 

# Tính gradient theo hướng y (đạo hàm riêng theo y) bằng mặt nạ Prewitt y
gradient_y = ndimage.convolve(image_array, np.array([[-1, -1, -1], [0, 0, 0], [1, 1, 1]])) # Ý nghĩa: -1 → giảm cường độ của pixel bên trái, 0 → pixel trung tâm không tác động, 1 → tăng cường độ pixel bên phải.
# Tính biên độ gradient
gradient_magnitude = np.sqrt(gradient_x**2 + gradient_y**2)

# Hiển thị ảnh
fig, axs = plt.subplots(2,1, figsize=(13, 18))

plt.subplot(2,1,1)
plt.imshow(gray_image, cmap='gray')
plt.title('Ảnh xám gốc')

plt.subplot(2,1,2)
plt.imshow(gradient_magnitude, cmap='gray')
plt.title('Ảnh xám gốc sau khi tính biên độ dộ dốc')

plt.show()
```
