```python
# Đọc một đường dẫn hình ảnh
my_path = '9bf15948-c9ac-41f6-abe4-38a5b155fb65-wm.png'
img = cv2.imread(my_path)

# In thông tin hình ảnh
print(img)  # 3 thông số [a b c] đại diện cho [red green blue]
```

```python
# Lấy kích thước của ảnh
img.shape # 3 thông số [a b c] trong .shape đại diện cho [row column dimemsion]
print(img.shape)
x, y, z = img.shape
x
y
z
print(x)
print(y)
print(z)
```

```python
# Hiển thị Hình ảnh
cv2.imshow('Logo_Drew', img)
# Chờ một khoảng thời gian rồi tắt
cv2.waitKey(10000) # Giải thích: 10000 là 10000 mili giây = 10 giây => Sau 10 giây ảnh sẽ tự động tắt!
# Đóng Cửa sổ của tab lại (Chỉ 1 cửa sổ tab)
cv2.destroyWindow('Logo_Drew')
```

## Color
### RGB(255, 0, 0) - Màu đỏ đậm
### RGB(0, 255, 0) - Màu xanh lá cây
### RGB(0, 0, 255) - Màu xanh dương
### RGB(255, 255, 0) - Màu vàng
### RGB(255, 0, 255) - Màu magenta
### RGB(0, 255, 255) - Màu cyan
### RGB(128, 128, 128) - Màu xám
### RGB(128, 0, 128) - Màu tím 
### RGB(255, 255, 255) - Màu trắng
### RGB(0, 0, 0) - Màu đen

```python
# Tách màu
b, g, r = cv2.split(img)
cv2.imshow('Logo_Drew', img)
cv2.imshow('Red', r)
cv2.imshow('Green', g)
cv2.imshow('Blue', b)

# Chờ một khoảng thời gian rồi tắt
cv2.waitKey(0) # Giải thích: 0 là 0 mili giây = 0 giây =>  Ảnh sẽ ko tự động tắt trừ khi mình tự tắt nó!
# Đóng Cửa sổ của tab lại (Chỉ 1 cửa sổ tab)
  # cv2.destroyWindow('Logo_Drew')
# Đóng tất cả cửa sổ tab
cv2.destroyAllWindows()
```

```python
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
image_2 = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # Hàm .cvtColor dùng để đổi màu sắc khác nhau
cv2.imshow('Logo_Drew', img)
cv2.imshow('gray_img', gray_img)
cv2.imshow('image_2', image_2)
# Chờ một khoảng thời gian rồi tắt
cv2.waitKey(0)
# Đóng tất cả cửa sổ tab
cv2.destroyAllWindows()
```

```python
# Thay đổi thông số của điểm ảnh
height, width, z = img.shape
quarter_h = height/2
quarter_w = width/2

# Green color
green_color = (0, 255, 0)

#Copy hình ảnh ra
img2 = img.copy()
# Thay đổi màu của góc 1/4 bên trái trên
for y in range(int(quarter_h)):
    for x in range(int(quarter_w)):
        img2[y, x] = green_color

# Hiển thị ra
cv2.imshow('Logo_Drew', img)
cv2.imshow('img2', img2)
# Chờ một khoảng thời gian rồi tắt
cv2.waitKey(0)
# Đóng tất cả cửa sổ tab
cv2.destroyAllWindows()
```

```python
# Các cách hiện thị tab hình ảnh (có thể reuse code)
display('Logo_Drew', img)     # Hiển thị cửa sổ tab 1
display('img2', img2)         # Hiển thị cửa sổ tab 2
display('gray_img', gray_img) # Hiển thị cửa sổ tab 3
display('image_2', image_2)   # Hiển thị cửa sổ tab 4
display('Red', r)             # Hiển thị cửa sổ tab 5
display('Green', g)           # Hiển thị cửa sổ tab 6
display('Blue', b)            # Hiển thị cửa sổ tab 7
```
