```python
# Giới thiệu thư viện Matplotlib
import matplotlib as mpl
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# Đường dẫn hình ảnh
my_path = 'Image.png'
```

```python
# Đọc đường dẫn hình ảnh
my_img = Image.open(my_path) #hoặc my_image = load_image(my_path)
display(my_img)
```

```python
# Đọc đường dẫn hình ảnh có hiển thị tọa độ x, y
my_img = Image.open(my_path) #hoặc my_image = load_image(my_path)
plt.imshow(my_img) #imshow là hàm có sẵn trong thư viện matplotlib
plt.show() #Dùng để hiển thị ảnh trong Matplotlib, có hoặc không cũng không sao
```

```python
# Vẽ đồ thị cơ bản
x = [1,2,3,4,5]
y = [10,15,20,30,5]
plt.plot(x, y)
plt.show()
```

```python
# Vẽ đồ thị cơ bản + thêm màu
x = [1,2,3,4,5]
y = [10,15,20,30,5]
plt.plot(x, y, color='red') # Thay đổi màu thành đỏ
plt.show()
```

```python
# Vẽ đồ thị cơ bản + thêm màu + tiêu đề
x = [1,2,3,4,5]
y = [10,15,20,30,5]
plt.plot(x, y, color='red')
plt.title('Ví Dụ về Biểu Đồ Đường') # Thêm tiêu đề
plt.show()
```

```python
# Tắt hệ trục tọa độ
x = [1,2,3,4,5]
y = [10,15,20,30,5]
plt.plot(x, y, color='red')
plt.title('Ví Dụ về Biểu Đồ Đường')
plt.axis('off') # Dùng để tắt hệ trục tọa độ, muổn mở dùng 'on'
plt.show()
```

```python
# Vẽ trực tiếp trên hình ảnh
#Đọc hình ảnh
my_image = Image.open(my_path)
x = [100,100,400,400]
y = [200,300,200,300]
plt.imshow(my_image) # Hiển thị hình ảnh
plt.plot(x,y, 'r*') # Đánh dấu các điểm, 'r*' viết tắt màu đỏ => hiển thị ra màu đỏ, có thể dùng màu khác thay thế.
plt.plot(x,y, 'bs:')
plt.show()
```
