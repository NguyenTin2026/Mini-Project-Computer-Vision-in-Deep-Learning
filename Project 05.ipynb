```python
# Import lớp Image 
from PIL import Image
from IPython.display import display
from imgtools import *
```

```python
# Đường dẫn hình ảnh
my_path = 'image.png'
```

```python
# Đọc hình ảnh
img = load_image(my_path)
```

```python
display(img)
```

```python
print(img.size)
```

```python
# Xoay hình ảnh
rotated_image = img.rotate(180) # hoặc rotated_image = rotated_image.rotate(90)
display(rotated_image)
```

```python
# Resize hình ảnh
img.resize((150,220))
```

```python
# Đảo chiều(Transpose) hình ảnh
transposed_image = img.transpose(Image.Transpose.ROTATE_90) #Có thể thay số 90 độ thành số khác
display(transposed_image)
```
