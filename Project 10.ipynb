```python
# Import Thư Viện
from PIL import Image
import matplotlib.pyplot as plt

# Đường dẫn file hình ảnh
my_path = 'c49f03e1-ecb8-4d05-bf16-90a20a54c0b1.png'
#Đọc file hình ảnh
img = Image.open(my_path)
# Chuyển đổi backend
plt.switch_backend('tkagg') #Giải thích: dùng Tkinter (ổn định, phổ biến) vừa vẽ hình và cửa sổ pop-up.
# Hiển thị hình ảnh
plt.imshow(img)
plt.title('Click on the image to select points')
# Sử dụng hàm ginput để chọn điểm trên hình ảnh
points = plt.ginput(6) # 6 là số điểm chọn trên 1 hình ảnh, có thể chọn số lượng tùy ý
print(points)
# Hiển thị ra
plt.show() 
```

```python
# Vẽ lại các điểm đã chọn bằng dấu * màu đỏ
plt.close()

# Vẽ lại
plt.imshow(img) # Dùng hàm này để nó show ra lại kết quả bạn vừa mới vẽ xong!
for point in points:
    x, y = point
    plt.plot(x,y, 'y*--')
plt.show()
```
