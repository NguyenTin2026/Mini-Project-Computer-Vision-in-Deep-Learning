```python
from PIL import Image
import numpy as np 
import matplotlib.pyplot as plt 
from scipy import ndimage

# Đọc hình ảnh
image_path = '9e9f8853-bced-436d-9faf-5c7d80bd010f.png'
img = Image.open(image_path)

# Chuyển sang ảnh xám
gray_image = img.convert('L')

#Chuyển ảnh xám về mảng
image_array = np.array(gray_image)

# Làm mờ ảnh xám với bộ lọc Gaussian
sigma_1 = 0.5
blurred_image_a = ndimage.gaussian_filter(image_array, sigma = 0.5, mode = 'constant')
blurred_image_b = ndimage.gaussian_filter(image_array, sigma = 3, mode = 'constant')
blurred_image_c = ndimage.gaussian_filter(image_array, sigma = 15, mode = 'constant')

# Hiển thị ảnh
fig, axs = plt.subplots(2,2, figsize = (10,15))
plt.subplot(2,2,1)
plt.imshow(gray_image, cmap='gray')
plt.title('Ảnh xám gốc')

plt.subplot(2,2,2)
plt.imshow(blurred_image_a, cmap='gray')
plt.title('Sigma=0.5')

plt.subplot(2,2,3)
plt.imshow(blurred_image_b, cmap='gray')
plt.title('Sigma=3')

plt.subplot(2,2,4)
plt.imshow(blurred_image_c, cmap='gray')
plt.title('Sigma=15')
```
