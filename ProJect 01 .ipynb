```python
# Import Hình ảnh vào
from PIL import Image # Sử dụng thư viện PIL
```

```python
# Thư mục chứa Hình ảnh
dir = "./Images/image01.png"
```

```python
# Đường dẫn đến Hình ảnh
image_path = "/.Images/image01.png"
print(image_path)
```

```python
# Sử dụng Image.open() để đọc Hình ảnh
image01 = Image.open(image_path)
```

```python
# Hiển thị hình ảnh
display(image01)
```

```python
# In một số thông tin
# In định dạng ảnh
print("Định dang ảnh: ", image01.format)
```

```python
# In ra size
print("Kích thước hình ảnh: ", image01.size)
```

```python
# Lưu ảnh và chuyển đổi ảnh
new_img_01_path = "/.Images/image01.png"
new_img_png_01_path = "/.Images/image01.png"
image01.save(new_img_01_path)
image01.save(new_img_png_01_path)
```

```python
# Đóng tập tin
image01.close()
```
