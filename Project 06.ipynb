```python
from PIL import Image
from IPython.display import display #để hiển thị ra hình ảnh
from imgtools import *
```

```python
# Đường dẫn tới hình ảnh
my_directory = 'Image.png'
```

```python
# Đọc đường dẫn hình ảnh
my_image = load_image(my_directory)
display(my_image) # hiển thị hình ảnh ra
print(my_image.size) # In ra kích cỡ của hình ảnh
```

```python
# Chuyển từ ảnh RGB(ảnh màu) sang ảnh xám
gray_image = my_image.convert('L') #Giá trị L là ảnh xám , 1 là ảnh đen-trắng, P là ảnh palatted
display(gray_image)
```

```python
# Phân RGB(ảnh màu) thành 3 tầng màu khác nhau
red_band, green_band, blue_band = my_image.split()
display(red_band)
display(green_band)
display(blue_band)
```

```python
# Trộn nhiều hình ảnh
merged_image = Image.merge('RGB', (red_band, green_band, blue_band))
display(merged_image)
```
