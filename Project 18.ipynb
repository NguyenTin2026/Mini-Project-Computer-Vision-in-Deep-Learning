```python
import numpy as np 
import matplotlib.pyplot as plt
from scipy import ndimage
from PIL import Image
```

```python
# Đọc Ảnh
image = Image.open('df31e507-9f4d-4fcd-b194-bc2e09db82f2.png')

# Chuyển sang ảnh xám
gray_image = image.convert('L')

image_array = np.array(gray_image)               
# Chuyển sang ảnh nhị phân
binary_image = 1*(image_array < 200)

# Đánh nhãn
labeled_image, num_features = ndimage.label(binary_image)

# Tạo lưới subplot với 3 dòng và 1 cột
fig, axs = plt.subplots(3,1, figsize = (50, 50))

plt.subplot(3, 1, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Ảnh xám')

plt.subplot(3, 1, 2)
plt.imshow(binary_image, cmap='gray')
plt.title('Ảnh xám')

plt.subplot(3, 1, 3)
plt.imshow(labeled_image, cmap='gray') # Thay thành giá trị khác: cmap='nipy_spectral')
plt.title('Ảnh đã đánh nhãn ({num_features} đối tượng)')

plt.show()
```

# Thực hiện opening để loại bỏ nhiễu và nối các đối tượng


```python
# Đọc Ảnh
image = Image.open('df31e507-9f4d-4fcd-b194-bc2e09db82f2.png')

# Chuyển sang ảnh xám
gray_image = image.convert('L')

image_array = np.array(gray_image)               
# Chuyển sang ảnh nhị phân
binary_image = 1*(image_array < 200)

# Thực hiện opening để loại bỏ nhiễu và nối các đối tượng
opened_image = ndimage.binary_opening(binary_image, structure=np.ones((9, 5)))

# Đánh nhãn
labeled_image, num_features = ndimage.label(binary_image)

# Tạo lưới subplot với 3 dòng và 1 cột
fig, axs = plt.subplots(4,1, figsize = (50, 50))
plt.subplot(4, 1, 1)
plt.imshow(gray_image, cmap='gray')
plt.title('Ảnh xám')

plt.subplot(4, 1, 2)
plt.imshow(binary_image, cmap='gray')
plt.title('Ảnh nhị phân')

plt.subplot(4, 1, 3)
plt.imshow(binary_image, cmap='gray')
plt.title('Ảnh sau opening')

plt.subplot(4, 1, 4)
plt.imshow(labeled_image, cmap='gray') # Thay thành giá trị khác: cmap='nipy_spectral')
plt.title(f"Ảnh đã đánh nhãn ({num_features} đối tượng)")

plt.show()
```
