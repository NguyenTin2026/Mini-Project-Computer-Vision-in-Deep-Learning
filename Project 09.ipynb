```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np 
```

```python
# Đường dẫn hình ảnh
my_directory = "fde9c5d5-4e81-4ae7-919e-bdfcb3ce0180.png"
```

```python
# Đọc file hình ảnh
img = Image.open(my_directory)
plt.imshow(img) # hoặc display(img) nếu có thư viện display
```

```python
# Chuyển sang ảnh xám
img_gray = img.convert('L')
plt.imshow(img_gray, cmap='gray')
```

```python
# Hiển thị ra mảng của hình ảnh bằng thư viện numpy
img_array = np.array(img_gray)
print(img_array)
```

```python
# Tạo ra một hình
plt.figure()
plt.hist(img_array.flatten(), bins=128)
plt.title('Biểu Đồ Histogram của Hình Ảnh')
plt.xlabel('Giá trị của Pixel')
plt.ylabel('Số lượng')
plt.show()
```

```python
# Làm nổi bật đường biên
plt.figure() 
plt.gray()
plt.contour(img_gray, origin='image', colors='k') # Hàm plt.contour() dùng để vẽ đường đẳng mức(contour lines)
# Có thể thêm khác ngoài origin và colors: levels, cmap, alpha, linewidths, linestyles, extent, labels.
plt.show()
```

```python
# Minh họa 1 hình ảnh khác:
 # Đường dẫn hình ảnh
my_path = 'e0a9a040-5eed-427d-b99b-6b1d35ea3b5f.png'
 # Đọc file hình ảnh
my_img = Image.open(my_path)
plt.imshow(my_img) # hoặc display(img) nếu có thư viện display
 # Chuyển sang ảnh xám
my_img_gray = my_img.convert('L')
plt.imshow(my_img_gray, cmap='gray')
# Hiển thị ra mảng của hình ảnh bằng thư viện numpy
my_img_array = np.array(my_img_gray)
print(my_img_array)
 # Tạo một hình vẽ mới
plt.figure()
 # Đặt màu sắc là xám
plt.gray()
 # Vẽ biểu đồ đường biên(contour) của ảnh
plt.contour(my_img_gray, origin='image')
 # Hiển thị hình ảnh
plt.show()
 # Tạo ra một hình
plt.figure()
plt.hist(my_img_array.flatten(), bins=128)
plt.title('Biểu Đồ Histogram của Hình Ảnh')
plt.xlabel('Giá trị của Pixel')
plt.ylabel('Số lượng')
plt.show()
```
